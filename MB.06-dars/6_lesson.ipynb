```python
df=df.drop(columns=['icao_code','iata_code','gps_code','local_code','wikipedia_link','keywords','home_link'])
```

```python
df.info()
```

```python
# Preprocessing
def toldir(df):
    for col in df.columns:
        if df[col].isnull().any():
            if df[col].dtype=='str':
                df[col]=df[col].fillna(df[col].mode()[0],inplace=True)
                   
            else:
                df[col]=df[col].fillna(df[col].mean(),inplace=True)
                
    return df            
```

```python
df=toldir(df)
df.info()
```

```python
df.nunique()
```

```python
# Encoding
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()

def encodla(df):
    for col in df.columns:
        if df[col].dtype=='str':
            if df[col].nunique()<5:
                dummies=pd.get_dummies(df[col],prefix=col,dtype=int)
                df=pd.concat([df.drop(columns=col),dummies],axis=1)
            else:
                df[col]=encoder.fit_transform(df[col])
    return df            

```

```python
df=encodla(df)
df.info()
```

```python
df.head()
```

```python
#scaling
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
def scaling(df):
    
    num_col=df.select_dtypes(include=['int64','float64']).columns.drop('elevation_ft')
    df[num_col]=scaler.fit_transform(df[num_col])
    return df    

    
```

```python
df=scaling(df)
df.head()
```

```python
import os
```

```python
output_folder='data/preprocessed'
os.makedirs(output_folder,exist_ok=True)
output_path=os.path.join(output_folder,'preprocessed_data_csv')
df.to_csv(output_path,index=False)
                    
```

```python
x=df.drop('elevation_ft',axis=1)
y=df['elevation_ft']
```

```python
#Training
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
from sklearn.linear_model import LinearRegression,Lasso,Ridge
```

```python
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
#prediction
y_pred=lr.predict(x_test)
y_pred[0]
```

```python
#Evaluation
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error,root_mean_squared_error
r2=r2_score(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
rmse=root_mean_squared_error(y_test,y_pred)
```

```python
print('r2:',r2)
print('mae:',mae)
print('mse:',mse)
print('rmse:',rmse)
```
