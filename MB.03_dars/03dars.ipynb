```python
#map+Lambda
list(map(lambda x:x*2, [1,2,3,4,5,6]))
```

```python
def qosh(x):
    return list(map(lambda x: x+3, [2,3,4,5]))
```

```python
qosh(x)
```

```python
#filter + Lambda
list(filter(lambda x:x>1, [1,2,3,4,5,-4,-7]))
```

```python
list(filter(lambda x:x>0 and x%2==0, [1,2,3,4,5,6,-4,-6]))
```
