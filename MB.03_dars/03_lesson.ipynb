```python
def makc(x,y):
    if x>y:
        print(x,' katta son')
    elif x<y:
        print(y,' katta son')
    else:
        print('ikkalasi teng')
        return x,y
```

```python
makc(45,56)
```

```python
makc(34,24)
```

```python
makc(45,45)
```
