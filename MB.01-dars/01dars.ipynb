# Visualiation 

```python
import pandas  as pd
```

```python
df=pd.read_csv('bank-.csv')
```

```python
df.info()
```

```python
df['Target']
```

```python
df['Target'].nunique()
```

```python
# Missing qiymatlar
def tozala(df):
    for col in df.columns:
        if df[col].isnull().any():
            if df[col].dtype=='str':
                df.fillna({col:df[col].mode()[0]},inplace=True)
            else:
                df.fillna({col:df[col].mean()},inplace=True)
    return (df)           
```

```python
df=tozala(df)
df.info()
```

```python
df.head()
```

```python
# Encoding
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
def encodla(df):
    for col in df.columns:
        if df[col].dtype=='str':
            df[col]=encoder.fit_transform(df[col])
    return (df)        
```

```python
df=encodla(df)
df.head()
```

```python
df.info()
```
