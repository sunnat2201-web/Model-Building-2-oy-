# Visualization


```python
import pandas as pd
df=pd.read_csv('student_mental_health_hybrid_learning.csv')
```

```python
df.info()
```

```python
def toldir(df):
    for col in df.columns:
        if df[col].isnull().any():
            if df[col].dtype=='str':
                df.fillna({col:df[col].mode()[0]},inplace=True)
            else:
                df.fillna({col:df[col].mean()},inplace=True)
    return df            
```

```python
df=toldir(df)
df.info()
```

```python
df['cgpa']
```

# Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,6))
plt.hist(df['age'],bins=20,color='blue',edgecolor='black')
plt.title('yosh')
plt.show
```

```python
plt.figure(figsize=(10,6)) #Figura o'lchami
plt.hist(df['cgpa'],bins=20,color='skyblue',edgecolor='black')
plt.title('gpa taqsimoti')
plt.ylabel('soni')
plt.xlabel('gpa darajasi')
plt.show
```
